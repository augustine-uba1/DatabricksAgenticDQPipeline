```python
import json

dbutils.widgets.text("sources_config_path", "")
cfg_path = dbutils.widgets.get("sources_config_path")

with open(cfg_path, "r", encoding="utf-8") as f:
    cfg = json.load(f)

keys = [s.get("key") for s in cfg.get("sources", []) if s.get("key")]
if not keys:
    raise ValueError(f"No sources found in config: {cfg_path}")

dbutils.jobs.taskValues.set(key="source_keys", value=keys)
print(f"Found {len(keys)} keys:", keys)
print("Config path used:", cfg_path)
```
